```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def SEM(std, n=200):
    return std / np.sqrt(n)
```

```python
data = np.load(r"METRICS_DHT_N100_ER_k10.0_log_beliefs-T_weights-msmoid_Snone_flex1_signal-bernoulli_gaussian-stds075_T200_200sims_1.npz")
print(data.files)

sem0 = SEM(data['T0_std'])
sem1 = SEM(data['T1_std'])
sem2 = SEM(data['T2_std'])
sem3 = SEM(data['T3_std'])

T0 = data['T0']
T1 = data['T1']
T2 = data['T2']
T3 = data['T3']

print(f"weights msmoid, flex=1, log-beliefs, bernoulli signal")

plt.figure(figsize=(9,6))
# plt.plot(T0, label=r'$\tau_0(t)$')
# plt.plot(T1, label=r'$\tau_1(t)$')
# plt.plot(T2, label=r'$\tau_2(t)$')
# plt.plot(T3, label=r'$\tau_3(t)$')
plt.errorbar(np.arange(len(T0)), T0, yerr=sem0, capsize=3, label=r'$\tau_0(t)$')
plt.errorbar(np.arange(len(T1)), T1, yerr=sem1, capsize=3, label=r'$\tau_1(t)$')
plt.errorbar(np.arange(len(T2)), T2, yerr=sem2, capsize=3, label=r'$\tau_2(t)$')
plt.errorbar(np.arange(len(T3)), T3, yerr=sem3, capsize=3, label=r'$\tau_3(t)$')
plt.xlabel(r'Timestep $t$', fontsize=24)
plt.ylabel(r"Total system belief $\tau_k(t)$", fontsize=24)
plt.legend(fontsize=20)
plt.yticks(np.arange(0, 1.1, 0.2), fontsize=20)
plt.xticks(np.arange(0, len(T0)+1, 25), fontsize=20)
plt.grid()
plt.xlim((0, len(T0)))
plt.ylim((0,1.02))
plt.savefig("old_vs_new_all_old_features.png", dpi=400, bbox_inches='tight')
plt.show()
```
